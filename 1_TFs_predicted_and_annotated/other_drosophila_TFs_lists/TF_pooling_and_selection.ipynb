```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib_venn import venn3
from upsetplot import from_contents, plot
from venn import venn

directory_TF= 'TF_diffrent_db_lists/'
#FlyTF database 
flyTF= pd.read_csv(directory_TF+'FlyTF.csv',  usecols=['symbol'],)
#FlyTF database 
jasparTF= pd.read_csv(directory_TF+'JASPAR_d_melanogaster_tf.csv',  usecols=['Name'], )
jasparTF.drop_duplicates(inplace=True)
#Redfly database 
redflyTF= pd.read_csv(directory_TF+'redfly_list.csv', usecols=['gene_name'],)
redflyTF.drop_duplicates(inplace=True)
redflyTF =redflyTF[redflyTF != 'Unspecified'].dropna()
#Flybase database 
flybase= pd.read_table(directory_TF+'FlyBase_TF.txt', usecols=['SYMBOL'])

#Zhu_2022 paper
#Zhu, H., Zhao, S. D., Ray, A., Zhang, Y., & Li, X. (2022). A comprehensive temporal patterning gene network in Drosophila medulla neuroblasts revealed by single-cell RNA sequencing. Nature communications, 13(1), 1-19.
Zhu_2022= pd.read_csv(directory_TF+'Zhu_2022.csv',usecols=['Symbol'])
Zhu_2022.drop_duplicates(inplace=True)

#Seroka_2022 paper
#Seroka, A., Lai, S. L., & Doe, C. Q. (2022). Transcriptional profiling from whole embryos to single neuroblast lineages in Drosophila. Developmental Biology.
Seroka_2022= pd.read_csv(directory_TF+'Seroka_2022.csv',usecols=['SYMBOL'])
Seroka_2022.drop_duplicates(inplace=True)

#Tribolium predicted TFs using deeep TFs
tribolium_predicted= pd.read_csv(directory_TF+'tribolium_output_true_annotated.csv', sep='\t', usecols=['gene_symbol'])
tribolium_predicted.drop_duplicates(inplace=True)

#Drosophila predicted TFs using deeep TFs
drosophila_predicted= pd.read_csv(directory_TF+'droso_output_true_annotated.csv', sep=',', usecols=['gene_name'])
drosophila_predicted.drop_duplicates(inplace=True)
```

```python
set1= set(flyTF.symbol)
set2= set(jasparTF.Name)
set3= set(redflyTF.gene_name)
set4= set(flybase.SYMBOL)
set5= set(Zhu_2022.Symbol)
set6= set(Seroka_2022.SYMBOL)
set7= set(tribolium_predicted.gene_symbol)
set8= set(drosophila_predicted.gene_name)
```

```python
sets = {
    'FlyTF (1052)': set1,
    'JasparTF (141)': set2,
    'Flybase (628)': set4,
    'RedflyTF (304)': set3
}
venn(sets, figsize=(10, 10), fontsize=11, legend_loc="upper left", cmap='tab20')

# Save as vector PDF
plt.savefig("1_venn_diagram_TFs_in_databases.pdf", format="pdf", bbox_inches="tight")

# Optional: show plot
plt.show()
```
